# Basics of numpy

```python
import numpy as np
```

### creating array from list

```python
arr_1d = np.array([1, 2, 3, 4 ,5])
print("arr_1d: ", arr_1d)

arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print("arr_2d: ", arr_2d)
```

### List vs numpy array

```python
# can stores different types of elements
py_list = [1, 2, 3, 4.3, 'Rohit', {"key": "value"}]
print("Python list multiplication: ", py_list*2)

# stores only one type of elements
np_array = np.array([1, 2, 3, 4]) 
print("numpy array multiplication: ", np_array*2)
```

### Creating array from scratch

```python
zeroes_1d = np.zeros(5)
zeroes_2d = np.zeros((3, 4))
print("1d Zeroes: ", zeroes_1d)
print("2d Zeroes: \n", zeroes_2d)

zeros = np.zeros(5, int) # zeros(n, datatype)
print("zeros: ", zeros)

ones = np.ones(5, int) # zeros(n, datatype)
print("ones: ", ones)

full = np.full((2,3), 7)
print("full: \n", full)

random = np.random.random((2, 3))
print("Random: \n", random)

randint = np.random.randint(10, 20) # generate integer random no. from 10 to 19
print("Random integer: ", randint)
```

### arange()

```python
sequence1 = np.arange(10) # arange(start, stop, stepsize)
sequence2 = np.arange(2, 12, 2)
sequence3 = np.arange(10, 1, -1)
sequence4 = np.arange(0, 11, 1.5)

print("Sequence1 : ", sequence1)
print("Sequence2 : ", sequence2)
print("Sequence3 : ", sequence3)
print("Sequence4 : ", sequence4)
```

### Attributes of an Array

```python
arr = np.array([[1, 2, 3], 
                [4, 5, 6]])

print("Dimension: ", arr.ndim)
print("Shape: ", arr.shape)
print("Size: ", arr.size)
print("dType: ", arr.dtype)
print("ItemSize: ", arr.itemsize)
print("nBytes: ", arr.nbytes)
```

### Vector operations

```python
# `shape` has to be same
a1  = np.array([[1, 2, 3], [4, 5, 6]])
a2 = np.array([[2, 2, 2], [1, -1, 2]])

print("Addition: \n", a1 + a2)
print("Subtraction: \n", a1 - a2)
print("Multiplication: \n", a1 * a2)
print("Division: \n", a1 / a2)
```

### Array Reshaping

```python
arr = np.arange(10)
print("Original Array: ", arr)

# the new array should have the same number of elements as in the original array
# returns a view, instead of copy
reshaped = arr.reshape((2, 5))
print("Reshaped Array: ", reshaped)
```

### Array Flattening

```python
arr = np.array([[1, 2, 3], 
                [4, 5, 6]])
print("Original Array: ", arr)

# returns a copy
flattend = arr.flatten()
print("Flatten Array: ", flattend)

# return view , instead of copy
raveled = arr.ravel()
print("Ravel Array: ", raveled)
```

### Array Transpose

```python
arr = np.array([[1, 2, 3], 
                [4, 5, 6]])
print("Original Array: ", arr)

# return view , instead of copy
transpose = arr.T
print("Transposed Array: ", transpose)
```

### View

```python
print("Memory Block : 1 2 3 4\n")
# Both are just different `views` of the `same memory block`.

arr = np.array([1, 2, 3, 4])
print("Array: ", arr)

reshaped = arr.reshape(2, 2)
print("Reshaped: ", reshaped)

# change original array
arr[0] = 100

print("\nAfter changing arr:")
print("Reshaped: ", reshaped)

# change reshaped array
reshaped[1][1] = 999

print("\nAfter changing reshaped:")
print("Array: ", arr)

# Checking Shared Memory
print("\nDo they share memory?")
print(np.shares_memory(arr, reshaped))
```
