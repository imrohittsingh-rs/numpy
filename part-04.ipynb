### Maths Functions in Numpy

```python
import numpy as np
```

```python
arr = np.array([10, 20, 30.5, -40])
print("Original Array: ", arr)

print("Sorted Array: ", np.sort(arr))

# take input in radians
print("Sin values: ", np.sin(arr))
print("Cos values: ", np.cos(arr))
print("Tan values: ", np.tan(arr))
print("Tan inverse values: ", np.arctan(arr))

print("Biggest element: ", np.max(arr))
print("Smallest element: ", np.min(arr))
print("Sum of all values: ", np.sum(arr))
print("Avg of all values: ", np.mean(arr))

print("Absolute: ", np.abs(arr))
print("Square root: ", np.sqrt(arr[:-1]))
print("Power: ", np.power(arr, 2))
print("log: ", np.log(np.abs(arr))) # natural log base e
print("exp: ", np.exp(arr))



```
